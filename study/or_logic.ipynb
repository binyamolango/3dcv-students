```python
import numpy as np
import matplotlib.pyplot as plt

# ==========================
# Neuron Parameters (OR Gate)
# ==========================
w1 = 1
w2 = 1
b = -0.5

# Decision Boundary:
# x1 + x2 - 0.5 = 0
# x2 = 0.5 - x1

x1 = np.linspace(0, 1.2, 500)
x2 = 0.5 - x1

# Create mesh grid
xx, yy = np.meshgrid(
    np.linspace(0, 1.2, 500),
    np.linspace(0, 1.2, 500)
)

# Neuron output before activation
zz = w1 * xx + w2 * yy + b

# Figure
plt.figure(figsize=(8, 8))

# Shade output regions
plt.contourf(
    xx,
    yy,
    zz,
    levels=[-1000, 0, 1000],
    colors=["#ffcccc", "#ccffcc"],
    alpha=0.5
)

# Plot decision boundary
plt.plot(
    x1,
    x2,
    color="black",
    linewidth=3,
    label=r"$x_1 + x_2 - 0.5 = 0$"
)

# OR gate input points
points = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

# Compute outputs
outputs = (
    w1 * points[:, 0]
    + w2 * points[:, 1]
    + b > 0
).astype(int)

# Plot points with labels
for p, out in zip(points, outputs):
    color = "green" if out == 1 else "red"

    plt.scatter(
        p[0],
        p[1],
        s=250,
        c=color,
        edgecolors="black",
        linewidth=2,
        zorder=5
    )

    plt.annotate(
        f"({p[0]}, {p[1]}) → {out}",
        (p[0], p[1]),
        xytext=(10, 10),
        textcoords="offset points",
        fontsize=10,
        fontweight="bold"
    )

# Axis settings
plt.xlim(-0.1, 1.2)
plt.ylim(-0.1, 1.2)

plt.xticks([0, 1])
plt.yticks([0, 1])

# Use LaTeX-style labels to avoid font warnings
plt.xlabel(r"$x_1$", fontsize=12)
plt.ylabel(r"$x_2$", fontsize=12)

plt.title("Single Neuron Implementing OR Logic", fontsize=14)
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend()

plt.tight_layout()
plt.show()
```
